```python
# The autoreload extension will automatically load in new code as you edit files, 
# so you don't need to restart the kernel every time
%load_ext autoreload
%autoreload 2
import os
from camera_calib import *

# Define paths.
DATA_PATH = os.path.join(os.getcwd(), 'images')  # Hoặc 'chessboard' tùy tên thư mục chứa ảnh của bạn

IMG_TO_USE = list(range(10, 36))
IMG_PATHS = [
    os.path.join(DATA_PATH, f'color_{i:02d}.png')
    for i in IMG_TO_USE
]
```

```python
# Compute world coordinate points
X_w, Y_w, Z_w = generate_chessboard_3D_world_coordinates()

# For each image, compute the pixel coordinates of the corners
u_meas = []; v_meas = []
for img_path in IMG_PATHS:
    u, v = generate_chessboard_2D_pixel_coordinates(img_path)
    u_meas.append(u); v_meas.append(v)

# Compute homography for each image.
homographies = []
for u, v in zip(u_meas, v_meas):
    homographies.append(compute_homography(u, v, X_w, Y_w))

# Compute intrinsics from homographies.
K = compute_intrinsics(homographies)
print("Intrinsics:")
print(K)
```
